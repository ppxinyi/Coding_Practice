We will use the term frequency - inverse document frequency (tf-idf) metric to rank the abstracts.
This metric is used in information retrieval to score the relevance of each document relative to the words in the search query.
Intuitively the term frequency measures if the word appears more times in the document that document which would make that document more relevant (adjusting for the document length).
The inverse document frequency accounts for the fact that some words are more common than others.
The formula is: 

```
tf_idf = tf * idf
tf = count of given term in document/ total number of terms in document
idf = log(Number of documents/ Number of documents with the given term)
```

````python
# simplest solution - works if you change the key to AC or AN 
# or take another variant line from a vcf file

# set your key
search_key = "DP"

# add "=" - that is the delimiter between key and value
search_key_eq = search_key + "="

# isolating/extracting the INFO column - that is everything after the last tab

# find the last tab - using rfind searches from the right - returns highest index
idx_tab = variant.rfind("\t")

# slicing to get the INFO part of the variant 
# start at the tab index + 1 so we do not include the tab
# stop at the end of the string - no stop index provided
info_col = variant[idx_tab + 1:]

# find the index of the key in the INFO string
# we search for the key + equal
idx_key = info_col.find(search_key_eq)

# find the index of the first ; after the key
# we give a start index - which is the key index, 
# otherwise we search from the beginning and find ; that are before the key (DP)
idx_end = info_col.find(";", idx_key)

# slice the INFO string from right after the equal sign until the ; 
# to the index of the key we add the length of the key and equal 
# to start right after the equal sign
# the stop index is the index of the first ; that is after the key
value = info_col[idx_key + len(search_key_eq): idx_end]

# make the value an int
DP = int(value)

# look at the value
DP
````
